```python
#121
import pandas as pd
import numpy as np
data = [
    ["037730", "3R", 1510, 7.36],
    ["036360", "3SOFT", 1790, 1.65],
    ["005760", "ACTS", 1185, 1.28],
]
columns = ["종목코드", "종목명", "현재가", "등락률"]
df = pd.DataFrame(data=data, columns=columns)
df.set_index('종목코드', inplace=True)
cond = df["현재가"]>=1500
df[cond]
```

```python
#122
cond2 = (df["현재가"] >= 1500) & (df["현재가"] < 1700)
df[cond2]
```
