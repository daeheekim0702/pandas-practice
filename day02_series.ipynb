#Day02-시리즈 기초

```python
import pandas as pd

s = pd.Series([45000,380000,1500000])
s
```

```python
print(s.values) # print로 감싸서 출력
s.index         # 마지막 줄이라 자동 출력
```

```python
s = pd.Series([45000,380000,1500000], index = ["사무용품","회식비","강사비"])
s
```

```python
s = pd.Series({"사무용품":45000,"회식비":380000,"강사비":1500000})
s
```

```python
print(s["회식비"]) # 이름으로 꺼내기
print(s.iloc[0])   # 위치(0번째)로 꺼내기
```

```python
print(s.iloc[0:2])        #0번째부터 2번째 직전까지 꺼내기
print(s[["회식비","강사비","사무용품"]])  #여러개는 대괄호 두개 (index도같이 나옴)
```

```python
print(s[s>=100000])
print(s.sum())
print(s * 1.1)
```

```python
index = ["메로나","누가바","빠삐코"]
data = [500,800,200]

s= pd.Series(data,index)
s
```
