```python
import pandas as pd

df = pd.DataFrame({
    "거래일": ["2026-03-02", "2026-03-07", "2026-03-15", "2026-03-21", "2026-03-06"],
    "적요": ["사무용품 구입", "회식비", "강사비 지급", "상품권 구입","유흥주점"],
    "금액": [45000, 380000, 1500000, 500000, 650000], 
})
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df["거래일"] = pd.to_datetime(df["거래일"])
df[df["금액"] >= 300000]
```

```python
df["요일"] = df["거래일"].dt.day_name()
df[df["요일"].isin(["Saturday", "Sunday"])]
```
