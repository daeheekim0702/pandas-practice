# 오답 정리 (day04 시리즈 복습)

## 반복된 실수 3가지
1. **철자 틀림** → 점(.)까지 치고 Tab 눌러서 자동완성으로 고르기
   - `pd.to_datetimes` ❌ → `pd.to_datetime` ⭕
   - `sort_value` ❌ → `sort_values` ⭕
   - `values_count` ❌ → `value_counts` ⭕
2. **점 앞의 시리즈 빠뜨림** → 항상 `시리즈.기능()` 모양
   - `count(s.isna())` ❌ → `s.isna().sum()` ⭕
   - `values_count()` ❌ → `v.value_counts()` ⭕
3. **다시 저장 안 함** → 메서드는 원본을 안 바꾸니까 `=`로 다시 담기
   - `amounts.str.replace(",", "")` ❌ → `amounts = amounts.str.replace(",", "")` ⭕

## 문제별 오답
- **#10 개수 세기**
  - `s.isna().sum()` = 빈칸 개수 (1)
  - `s.count()` = 빈칸 빼고 개수 (5)
  - `len(s)` = 빈칸 포함 전체 (6)
- **#12 평균**: `sum/len(s)`는 빈칸까지 세서 틀림 → `s2.mean()` 쓰기
  - 변수 이름을 `sum`으로 짓지 않기 (파이썬 기본 함수 덮어씀)
  - 큰 순서 정렬은 `ascending=False`
- **#13 날짜 거르기**: 값이 아니라 인덱스를 비교
  - `cond2 = s2.index > "2026-03-15"`
- **#14 strftime 에러**: 인덱스가 날짜일 때만 가능 → 셀 실행 순서 확인, 헷갈리면 Restart 후 Run All
- **#15**: 문제에서 쓰라는 변수(v)를 확인하고, 고유값 목록은 `v.unique()`
- **#17 엑셀 읽기**: 쉼표 없이 대괄호를 바로 붙이고, 열 이름은 시리즈 이름("지출액")
  - `pd.read_excel("review.xlsx", index_col=0)["지출액"]`
