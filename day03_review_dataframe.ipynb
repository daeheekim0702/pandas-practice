### 07. 시리즈 기본 메소드

```python
from pandas import Series 

data = {
    "2020-12-11": 73400,
    "2020-12-10": 72900,
    "2020-12-09": 73900,
    "2020-12-08": 71700,
    "2020-12-07": 72900
}
s = Series(data)
s.sum()
```

```python
from pandas import Series 

data = {
    "2020-12-11": 73400,
    "2020-12-10": 72900,
    "2020-12-09": 73900,
    "2020-12-08": 71700,
    "2020-12-07": 72900
}
s = Series(data)
s.mean()
```

```python
from pandas import Series 

data = {
    "2020-12-11": 73400,
    "2020-12-10": 72900,
    "2020-12-09": 73900,
    "2020-12-08": 71700,
    "2020-12-07": 72900
}

s = Series(data)
print(s.max())
print(s.min())
```

```python
from pandas import Series #상승률 계산하는 문제 s.pct_change()

data = {
    "2020-12-11": 73400,
    "2020-12-10": 72900,
    "2020-12-09": 73900,
    "2020-12-08": 71700,
    "2020-12-07": 72900
}

s = Series(data)
s1 = s.shift(1)
print((s-s1)/s1)
print(s.pct_change())
```

```python
from pandas import Series 

data = {
    "2020-12-11": 1.01,
    "2020-12-10": 0.99,
    "2020-12-09": 1.03,
    "2020-12-08": 1.02,
    "2020-12-07": 1
}

s = Series(data)
print(s.product())
```

s.sum()       # 전부 더하기
s.product()   # 전부 곱하기
s.mean()      # 평균
s.cumprod()   # 누적해서 곱하기 (하루하루 누적 수익률 보기)

### 08. 시리즈 데이터 변경

```python
import numpy as np
from pandas import Series 

data = [1, 2, 3, 4, np.nan]
s = Series(data=data)
s.count()
```

- s.count() : 빈칸 빼고 개수 / len(s) : 전체 개수 / s.isna().sum() : 빈칸 개수

```python
data = ["100", "200", "300"]
index = ["철수", "영희", "아름"]
s = Series(data, index)
s=s.astype("int64")
s
```

```python
data = ["100,000", "200,000", "300,000"]
index = ["철수", "영희", "아름"]
s = pd.Series(data, index)
s = s.str.replace(',','')
s = s.astype('int32')
s # 다시풀어보기
```

```python
s = pd.Series(["30세", "40세", "50세"], index=["김대리", "이과장", "박차장"])
s = s.str.replace("세",'')
s = s.astype('int32')
s
```
